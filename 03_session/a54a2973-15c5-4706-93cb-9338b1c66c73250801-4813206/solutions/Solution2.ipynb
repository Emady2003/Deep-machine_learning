### Set PassengerId as the index 

```python
df_titanic_indexed = df_titanic.set_index('PassengerId')
df_titanic_indexed
```

### Create a pie chart presenting the male/female proportion

```python
sex = df_titanic['Sex'].unique()
quantity = df_titanic['Sex'].value_counts()

data = pd.Series(quantity, index=sex)
data.plot.pie(autopct='%1.0f%%')
```

### Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
colors = {'male': 'blue', 'female': 'red'}
df_titanic.plot.scatter(x='Fare', y='Age', s = 8, c=df_titanic['Sex'].map(colors))

plt.show()
```

### How many people survived?

```python
df_titanic['Survived'].sum()
```

### Create a histogram and box plot with the Fare payed (use subplot)

```python
fig, axes = plt.subplots(2, 1, figsize=(8, 6), sharex=True, gridspec_kw={'height_ratios': [0.2, 0.8]})
df_titanic.boxplot(column='Fare', ax=axes[0], vert=False)
axes[0].set_title('Box Plot of Fare')
axes[0].set_xlabel('')

df_titanic.hist(column='Fare', ax=axes[1], bins=10, edgecolor='black')
axes[1].set_title('Histogram of Fare')
axes[1].set_xlabel('Fare')
axes[1].set_ylabel('')
plt.show()
```

### Drop passengerID and Ticket columns

```python
df_titanic.drop(columns=['PassengerId', 'Ticket'])
```

### Is there a meaningful relationship between living and dying with age? (show relationship by visual representation)

```python
df_titanic.boxplot(column='Age', by='Survived')
plt.title('Box Plot of Age based on survival')
plt.suptitle('') # Suppress the default suptitle generated by 'by'
plt.xlabel(' ')
plt.ylabel('Age')
#plt.xticks()
plt.xticks([1, 2], ['Survived', 'Not Survived'])
plt.show()
```

### Tell me about the relationship between the age, Pclass and sex

```python
colors = ['deeppink' if val == 'female' else 'blue' for val in df_titanic['Sex']]
df_titanic.plot.scatter(x='Age', y='Pclass', s = 8, c = colors, figsize=(8, 2), xlim = [-4, 100], ylim = [0.5, 3.5])

plt.xticks([10, 20, 30, 40, 50, 60, 70, 80])
plt.yticks([1, 2, 3], ['1', '2', '3'])
plt.show()
```

```python
df_titanic.plot.scatter(x='Age', y='Pclass', s = 8, c = colors, figsize=(8, 2), xlim = [-4, 100], ylim = [0.5, 3.5])
plt.scatter(df_titanic[df_titanic['Sex'] == 'male']['Age'], 
            df_titanic[df_titanic['Sex'] == 'male']['Pclass'],
            s = 8, 
            c = 'blue')
plt.scatter(df_titanic[df_titanic['Sex'] == 'female']['Age'],
             df_titanic[df_titanic['Sex'] == 'female']['Pclass'],
             s = 8, 
             c = 'pink')
plt.xticks([10, 20, 30, 40, 50, 60, 70, 80])
plt.yticks([1, 2, 3], ['1', '2', '3'])
plt.show()
```
